```python
import os
import yfinance as yf
import pandas as pd
from fredapi import Fred
from tqdm import tqdm

ROOT="Sem Project"

START="2017-07-01"
END="2023-07-01"

os.makedirs(f"{ROOT}/Commodity_Data",exist_ok=True)
os.makedirs(f"{ROOT}/Forex_Data",exist_ok=True)
os.makedirs(f"{ROOT}/Crypto_Data",exist_ok=True)
os.makedirs(f"{ROOT}/Bond_Data",exist_ok=True)

# ----------------------
# COMMODITIES
# ----------------------

commodities={
"Gold":"GC=F",
"Silver":"SI=F",
"Copper":"HG=F",
"Platinum":"PL=F",
"Palladium":"PA=F",
"Brent":"BZ=F",
"WTI":"CL=F",
"NaturalGas":"NG=F",
"HeatingOil":"HO=F",
"RBOB":"RB=F",
"Corn":"ZC=F",
"Wheat":"ZW=F",
"Soybeans":"ZS=F",
"Sugar":"SB=F",
"Cotton":"CT=F"
}

print("Downloading commodities")

for name,ticker in tqdm(commodities.items()):

    df=yf.download(ticker,start=START,end=END,auto_adjust=False,progress=False)

    if len(df):
        df.to_csv(f"{ROOT}/Commodity_Data/{name}.csv")

# ----------------------
# FOREX
# ----------------------

forex={
"AUD":"AUDUSD=X",
"CAD":"USDCAD=X",
"CHF":"USDCHF=X",
"CNY":"USDCNY=X",
"EUR":"EURUSD=X",
"GBP":"GBPUSD=X",
"INR":"USDINR=X",
"JPY":"USDJPY=X",
"MXN":"USDMXN=X",
"NZD":"NZDUSD=X",
"RUB":"USDRUB=X",
"SGD":"USDSGD=X",
"TRY":"USDTRY=X",
"ZAR":"USDZAR=X",
"BRL":"USDBRL=X",
"USDI":"DX-Y.NYB"
}

print("Downloading forex")

for name,ticker in tqdm(forex.items()):

    df=yf.download(ticker,start=START,end=END,auto_adjust=False,progress=False)

    if len(df):
        df.to_csv(f"{ROOT}/Forex_Data/{name}.csv")

# ----------------------
# CRYPTO
# ----------------------

crypto={
"BTC":"BTC-USD",
"ETH":"ETH-USD",
"USDT":"USDT-USD",
"BNB":"BNB-USD",
"XRP":"XRP-USD",
"ADA":"ADA-USD",
"SOL":"SOL-USD",
"DOGE":"DOGE-USD",
"TRX":"TRX-USD",
"LTC":"LTC-USD",
"LINK":"LINK-USD",
"XLM":"XLM-USD",
"XMR":"XMR-USD",
"EOS":"EOS-USD",
"ETC":"ETC-USD"
}

print("Downloading crypto")

for name,ticker in tqdm(crypto.items()):

    df=yf.download(ticker,start=START,end=END,auto_adjust=False,progress=False)

    if len(df):
        df.to_csv(f"{ROOT}/Crypto_Data/{name}.csv")



print("Done.")
```

```python
import pandas as pd
import glob
import os
import numpy as np

folders = ["Crypto_Data", "Forex_Data", "Commodity_Data"]

def log_return(df):
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    df = df.sort_values("Date").reset_index(drop=True)
    df = df.sort_values("Date")
    df["Return"] = 100 * np.log(df["Price"] / df["Price"].shift(1))
    return df[["Date", "Return"]]

for folderX in folders:

    folder = os.path.join("Data", folderX)
    return_folder = os.path.join("Returns Data", folderX)
    os.makedirs(return_folder, exist_ok=True)

    files = glob.glob(os.path.join(folder, "*.csv"))

    for file in files:

        df = pd.read_csv(
            file,
            header=0,
            skiprows=[1, 2]
        )

        df.rename(
            columns={df.columns[0]: "Date"},
            inplace=True
        )

        if "Adj Close" in df.columns:
            price_column = "Adj Close"
        elif "Close" in df.columns:
            price_column = "Close"
        elif "Price" in df.columns:
            price_column = "Price"
        else:
            print(f"Skipping {file}: No suitable price column found.")
            continue

        df = df[["Date", price_column]]
        df.rename(columns={price_column: "Price"}, inplace=True)

        df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
        df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

        df = df.dropna(subset=["Date", "Price"])

        returns = log_return(df)

        filename = os.path.basename(file)

        returns.to_csv(
            os.path.join(return_folder, filename),
            index=False
        )

#stocks

folder = os.path.join("Data", "Stock_Data")
return_folder = os.path.join("Returns Data", "Stock_returns")

os.makedirs(return_folder, exist_ok=True)

files = glob.glob(os.path.join(folder, "*.csv"))

for file in files:

    df = pd.read_csv(file, thousands= ',')

    df = df[["Date", "Price"]]

    df = df.dropna(subset=["Date", "Price"])

    returnsstock = log_return(df)

    filename = os.path.basename(file)

    returnsstock.to_csv(
        os.path.join(return_folder, filename),
        index=False
    )

#bonds 
bond_folder = "Data/Bond_Data"
bond_return_folder = "Returns Data/Bond_returns"

os.makedirs(bond_return_folder, exist_ok=True)

def bond_return(df):
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
    df = df.sort_values("Date").reset_index(drop=True)
    y = df["Price"] / 100
    df["Return"] = 100 * np.log(
        (10 *y.shift(1) + 1) /
        (10 *y + 1)
    )
    return df[["Date", "Return"]]

files = glob.glob(os.path.join(bond_folder, "*.csv"))

for file in files:

    df = pd.read_csv(file)

    df = df[["Date", "Price"]]

    returns = bond_return(df)

    filename = os.path.basename(file)

    returns.to_csv(
        os.path.join(bond_return_folder, filename),
        index=False
    )

print("All return datasets created successfully.")
```
